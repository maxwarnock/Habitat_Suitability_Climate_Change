## The Great Basin Bristlecone Pine

The Great Basin Bristlecone Pine (*Pinus longaeva*) is known for being one of the longest living tree species on Earth. These trees live in harsh conditions at high elevation (about 7,200 to 12,000 ft across their range, and mostly above 8,000 ft in Nevada) in Nevada, California, and Utah ([US Forest Service](https://research.fs.usda.gov/feis/species-reviews/pinlon)). They are incredibly slow growing, and some years, they do not even add rings to their trunks. One tree in Great Basin National Park, known as Prometheus, was estimated to be 4,700 to 5,200 years old when it was cut down in 1964 ([National Park Service](https://www.nps.gov/grba/planyourvisit/identifying-bristlecone-pines.htm)).

Close relatives include the Rocky Mountain Bristlecone Pine (*Pinus aristata*) and the Foxtail Pine (*Pinus balfouriana*), which are separate species and do not live nearly as long as Great Basin Bristlecone Pines ([National Park Service](https://www.nps.gov/grba/planyourvisit/identifying-bristlecone-pines.htm)).

### Bristlecone Pine Habitat

The Great Basin Bristlecone Pine occurs commonly in thin, rocky substrates, usually derived from limestone and dolomite. These soils are alkaline, high in calcium and magnesium, and low in phosphorus, meaning the pH values of these soils are generally above 7 (basic). Mean precipitation is about 12 inches per year in the White Mountains of California, and about twice that in the Snake Range of Nevada. The tree is more prolific on south and west facing slopes but can also be found on other slope directions, and it generally occurs on slopes ranging from 10 to 50 percent ([US Forest Service](https://research.fs.usda.gov/feis/species-reviews/pinlon)).

### Bristlecone Pines and Climate Change

Due to warming global temperatures, I'm interested in understanding if and how the range of the Great Basin Bristlecone Pine could be affected. This tree thrives at high elevations where little other vegetation grows, which limits the fuel available for wildfires ([National Park Service](https://www.nps.gov/grba/planyourvisit/identifying-bristlecone-pines.htm)). However, future changes in temperature and precipitation may also impact the ability of this species to survive in its current habitats. The US Forest Service notes that some populations on marginal sites in the interior Great Basin are already threatened by climate change ([US Forest Service](https://research.fs.usda.gov/feis/species-reviews/pinlon)). Therefore, I created habitat suitability maps to predict the Bristlecone Pine's future habitat suitability range for two sites in Nevada.

## Study Sites

I selected two sites where Great Basin Bristlecone Pines are common. The first site is [Great Basin National Park](https://www.nps.gov/grba/index.htm) (GBNP) in east-central Nevada, near the Utah border. The other site is a portion of [Humboldt-Toiyabe National Forest](https://www.fs.usda.gov/r04/humboldt-toiyabe) (HTNF) in the Spring Mountains, just west of Las Vegas. Humboldt-Toiyabe covers a lot of different areas, so I picked its southernmost section to compare with the northern site. This allows for comparison of how habitat suitability is changing in the northern and southern parts of the state for this species.

## Habitat Suitability Criteria

| Criterion | Suitable range |
|---|---|
| Soil pH | above 7.0 |
| Elevation | 8,000 - 12,500 ft (2,440 - 3,810 m) |
| Aspect | South (180°) or west (270°) preferred; other aspects weighted down to 0.7 |
| Slope | 10 - 50 percent (about 6 - 27 degrees) |
| Precipitation | 30 - 95 cm per year (12 - 37 inches) |
| Max average annual temperature | 10.0 - 18.0 °C |
| Min average annual temperature | -7.8 - 4.0 °C |

The soil, aspect, and slope criteria come directly from the [US Forest Service](https://research.fs.usda.gov/feis/species-reviews/pinlon). The precipitation range runs from the driest reported sites (about 12 inches, or 30 cm, per year in the White Mountains; [US Forest Service](https://research.fs.usda.gov/feis/species-reviews/pinlon)) to the wettest (up to 37 inches, or 94 cm, per year on bristlecone pine sites in central Nevada; [Natural Resources Conservation Service](https://edit.sc.egov.usda.gov/catalogs/esd/028B/F028BY106NV)). The temperature ranges are approximations built around the reported averages (an average maximum temperature near 60°F, or 16°C, and an average minimum temperature near 18°F, or -8°C), widened to allow for variation across the species' range.

## Fuzzy Logic Model

All of the data layers were clipped and aligned to the same grid, using the elevation data as the reference. A fuzzy logic model then turned each criterion into a score from 0 to 1. A score of 1 means the pixel is fully inside the suitable range, and the score drops gradually toward 0 the further a pixel is outside it, instead of switching straight from suitable to unsuitable. The scores were multiplied together, pixel by pixel, to get a final habitat suitability score.

Elevation, slope, and soil do most of the work in the early century. The climate scores are close to 1 across the whole park, because today's temperature and precipitation there fit the species well.

## Choosing Climate Models

I ran the habitat suitability model across four different climate models, one for each of these categories. For each site, I used the scatter plot feature on [Climate Toolbox](https://climatetoolbox.org/) to identify an appropriate model for each category, under a lower emissions scenario (RCP 4.5). All of these models project warming under RCP 4.5, so "Warm" and "Cooler" describe each model relative to the other models (more or less warming), not relative to today's climate.

| Category | GBNP | HTNF |
|---|---|---|
| Warm and Dry | MIROC-ESM | MIROC-ESM |
| Warm and Wet | HadGEM2-ES365 | CanESM2 |
| Cooler and Dry | inmcm4 | GFDL-ESM2M |
| Cooler and Wet | CNRM-CM5 | CNRM-CM5 |

Each model was averaged over two periods: early century (2006-2035) and late century (2071-2099).

## Discussion

The early century habitat suitability maps all indicate a strong level of habitat suitability (above 0.5) for the areas that overlap with the majority of occurrence data points for the species. This is expected for the early century, since Great Basin bristlecone pines are known to be very tolerant of harsh climate conditions ([Whitebark Pine Ecosystem Foundation](https://whitebarkfound.org/five-needle-pines/great-basin-bristlecone-pine/)), and the values for early century temperature and precipitation are consistent with the species' habitat requirements.

The late century maps also indicate a strong level of habitat suitability for the areas overlapping with the species occurrence data points. However, the area of suitable habitat shrinks slightly in some of the climate models, in particular the MIROC-ESM maps for both sites. MIROC-ESM is a climate model predicting a warmer and drier climate compared to the other models, so it makes sense that it would predict the largest reduction in habitat range:

| Site | Area above 0.5 suitability, early century | Late century change |
|---|---|---|
| GBNP | about 207 km² | 0% to -9% (largest for MIROC-ESM) |
| HTNF | about 270 km² | -2% to -14% (largest for MIROC-ESM) |

Overall, the fact that habitat suitability is not greatly affected for the bristlecone pine is good news for this species. This is assuming that carbon emissions follow a trajectory similar to the RCP 4.5 emissions scenario. However, an RCP 8.5 or "business as usual" scenario ([Carbon Brief](https://www.carbonbrief.org/explainer-the-high-emissions-rcp8-5-global-warming-scenario/)) would likely impact habitat suitability more drastically. This analysis could easily be adapted to include RCP 8.5 climate model data.

Great Basin National Park, the northern site, has a larger buffer of suitable habitat surrounding the occurrence data than the Humboldt-Toiyabe National Forest site in southern Nevada. This appears to be largely due to the elevation requirements of the species. Roughly 80% of the Great Basin National Park site lies above 8,000 ft, compared to less than 20% of the Humboldt-Toiyabe site.

## Conclusion

This workflow shows how to create a habitat suitability map for a species, and allows for the comparison of different sites and climate models. The maps identified suitable habitat where the species currently lives (GBIF), which is promising evidence that the model works. The late century maps show a slight decrease in habitat range for the species, especially for the MIROC-ESM (warm and dry) climate model. However, the overall range of habitat suitability for the species remains intact through these late century predictions, assuming an RCP 4.5 emissions scenario.

## Limitations

- The habitat thresholds and tolerance ranges in the fuzzy logic model are approximations based on a small number of sources, and the results are sensitive to these choices
- The temperature tolerances are wide (15°C on either side of the optimal range), so warming lowers suitability gradually but never rules out a location entirely
- The MACA climate data has a resolution of about 4 km, which is much coarser than the 90 m elevation data, so it can't capture small-scale climate differences in steep terrain
- The model only looks at whether conditions suit the species in general, not whether new seedlings can establish
- Only the RCP 4.5 emissions scenario was modeled
